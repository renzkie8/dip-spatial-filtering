MEAN FILTERING

```python
# Load the noise image
img = cv2.imread('art.jpg')
#img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)  # Convert to RGB for proper display

# Generate random Gaussian noise
mean = 0
stddev = 50
noise = np.zeros(img.shape, dtype=np.int16)
cv2.randn(noise, mean, stddev)

# Add noise and clip to valid range [0, 255]
noisy_img = np.clip(img.astype(np.int16) + noise, 0, 255).astype(np.uint8)

# Apply Mean Filtering (Kernel Size 5x5)
filtered_img = cv2.blur(noisy_img, (5, 5))  # Mean filter with 5x5 kernel

# Display original, noisy, and filtered images
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

# Original Image
axes[0].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
axes[0].set_title("Original Image")
axes[0].axis("off")

# Noisy Image
axes[1].imshow(cv2.cvtColor(noisy_img, cv2.COLOR_BGR2RGB))
axes[1].set_title("Image w/ Noise")
axes[1].axis("off")

# Filtered Image (Mean Filtering)
axes[2].imshow(cv2.cvtColor(filtered_img, cv2.COLOR_BGR2RGB))
axes[2].set_title("Filtered Image (Mean Filtering)")
axes[2].axis("off")

plt.show()
```

```python
# Load the noise image
image = cv2.imread("art.jpg")
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)  # Convert for display

# Generate random Gaussian noise
mean = 0
stddev = 50
noise = np.zeros(img.shape, dtype=np.int16)
cv2.randn(noise, mean, stddev)

# Add noise and clip to valid range [0, 255]
noisy_img = np.clip(img.astype(np.int16) + noise, 0, 255).astype(np.uint8)

# Apply Median Filtering (Kernel Size 5x5)
filtered_img = cv2.medianBlur(noisy_img, 5)  # Kernel size must be odd (e.g., 3, 5, 7)

# Display original, noisy, and filtered images
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

# Original Image
axes[0].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
axes[0].set_title("Original Image")
axes[0].axis("off")

# Noisy Image
axes[1].imshow(cv2.cvtColor(noisy_img, cv2.COLOR_BGR2RGB))
axes[1].set_title("Image w/ Noise")
axes[1].axis("off")

# Filtered Image (Median Filtering)
axes[2].imshow(cv2.cvtColor(filtered_img, cv2.COLOR_BGR2RGB))
axes[2].set_title("Filtered Image (Median Filtering)")
axes[2].axis("off")

plt.show()
```

COMPARING AND CONTRASTING THE MEAN FILTER AND MEDIAN FILTER

As we can see in the ouput of each the Mean filter takes two arguments function which is the image and the kernel size for filtering the image. It first padded the input or selected image which it calculates the height and width of the image selected. The input image which is initialize a padded version been used a “cv2.copyMakeBorder” to mange the pixel of its border and prevent errors near image edges. 

The Median filter on the other hand takes two argument also takes two arguments: the image and the kernel size . In the output image we can see that the pixel value of the original image is replaced eliminating its noise, making its filter pixel non-linear. Unlike the Mean filter, the Median filter does not use an averaging method; instead, it sorts the pixel values within the kernel window and replaces the central pixel with the median of those values. This process makes the Median filter highly effective in removing salt-and-pepper noise while preserving edges better than the Mean filter. 

WITH THE ORIGINAL IMAGE W/O noise application of MEAN AND MEDIAN FILTERED IMAGES

```python
# with the original iamge
image = cv2.imread("art.jpg", cv2.IMREAD_COLOR)

# Define kernel size for filtering
kernel_size = (5, 5)  # 5x5 kernel for both filters

# Apply Mean Filter (Averaging Filter)
mean_filtered = cv2.blur(image, kernel_size)

# Apply Median Filter separately to each channel
median_filtered = np.zeros_like(image)
for i in range(3):  # Iterate over BGR channels
    median_filtered[:, :, i] = cv2.medianBlur(image[:, :, i], kernel_size[0])

# Create a large side-by-side display
plt.figure(figsize=(20, 10))

# Mean Filtered Image
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(mean_filtered, cv2.COLOR_BGR2RGB))
plt.title("Mean Filtered Image", fontsize=20)
plt.axis("off")

# Median Filtered Image
plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(median_filtered, cv2.COLOR_BGR2RGB))
plt.title("Median Filtered Image", fontsize=20)
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Load the image in grayscale
image_path = "art.jpg"  # Ensure the correct file path
img_gray = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

# Apply Gaussian Blur (optional, helps with noise reduction)
img_gray = cv2.GaussianBlur(img_gray, (5, 5), sigmaX=0.1, sigmaY=0.1)

# Apply the Laplacian filter
ddepth = cv2.CV_64F  # Use a high-depth format to prevent overflow
laplacian = cv2.Laplacian(img_gray, ddepth, ksize=3)

# Convert the Laplacian result to 8-bit scale for proper visualization
laplacian_abs = cv2.convertScaleAbs(laplacian)

# Display Original and Laplacian Filtered Image side by side
plt.figure(figsize=(15, 7))

# Original Grayscale Image
plt.subplot(1, 2, 1)
plt.imshow(img_gray, cmap="gray")
plt.title("Original Grayscale Image", fontsize=18)
plt.axis("off")

# Laplacian Filtered Image
plt.subplot(1, 2, 2)
plt.imshow(laplacian_abs, cmap="gray")
plt.title("Laplacian Filtered Image", fontsize=18)
plt.axis("off")

plt.tight_layout()
plt.show()
```

Apply Sobel filters (both horizontal and vertical) and compute the gradient magnitude

We can approximate the derivative in the X or Y direction  using the Sobel function, here are the parameters:

<p><code>src</code>: input image</p>
<p><code>ddepth</code>: output image depth, see combinations; in the case of 8-bit input images it will result in truncated derivatives</p>
<p><code>dx</code>: order of the derivative x</p>
<p><code>dx</code>: order of the derivative y</p>
<p><code>ksize</code> size of the extended Sobel kernel; it must be 1, 3, 5, or 7</p>


APPLYTING THE PARAMETERS OF THE FUNCTIONS

```python
ddepth = cv2.CV_16S
grad_x = cv2.Sobel(src=image_gray, ddepth=ddepth, dx=1, dy=0, ksize=5)
```

IN VERTICAL DIRECTION (Y)

```python
grad_y = cv2.Sobel(src=image_gray, ddepth=ddepth, dx=0, dy=1, ksize=5)
plt.imshow(grad_y, cmap='gray')
```

```python
# Load the original image in grayscale
img_gray = cv2.imread("art.jpg", cv2.IMREAD_GRAYSCALE)

# Apply Gaussian Blur to smooth the image and reduce noise
img_gray = cv2.GaussianBlur(img_gray, (3, 3), sigmaX=0.1, sigmaY=0.1)

# Apply Sobel filters (Horizontal and Vertical)
ddepth = cv2.CV_64F  # Use float64 to prevent overflow
grad_x = cv2.Sobel(src=img_gray, ddepth=ddepth, dx=1, dy=0, ksize=5)  # Sobel X
grad_y = cv2.Sobel(src=img_gray, ddepth=ddepth, dx=0, dy=1, ksize=5)  # Sobel Y

# Compute the Gradient Magnitude using the formula sqrt(SobelX^2 + SobelY^2)
grad_magnitude = np.sqrt(grad_x**2 + grad_y**2)

# Normalize to 8-bit scale for display
grad_magnitude = cv2.convertScaleAbs(grad_magnitude)

# Display the results side by side
plt.figure(figsize=(20, 10))

# Original Grayscale Image
plt.subplot(1, 3, 1)
plt.imshow(img_gray, cmap="gray")
plt.title("Original Grayscale Image", fontsize=18)
plt.axis("off")

# Sobel Gradient Magnitude 
plt.subplot(1, 3, 2)
plt.imshow(grad_magnitude, cmap="gray")
plt.title("Gradient Magnitude (Sobel)", fontsize=18)
plt.axis("off")

# Sobel X and Y Combined 
plt.subplot(1, 3, 3)
plt.imshow(cv2.convertScaleAbs(grad_x) + cv2.convertScaleAbs(grad_y), cmap="gray")
plt.title("Sobel X + Sobel Y", fontsize=18)
plt.axis("off")

plt.tight_layout()
plt.show()
```

b). Analyze and compare the results of the Laplacian and Sobel edge detections.

```python
# Load the image in grayscale mode
image_path = "art.jpg"  # Replace with your image path
img_gray = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)


# Apply Gaussian Blur to reduce noise
img_gray = cv2.GaussianBlur(img_gray, (3, 3), sigmaX=1, sigmaY=1)

# Apply Laplacian Filter (Second-order derivative for edge detection)
laplacian = cv2.Laplacian(img_gray, cv2.CV_64F, ksize=3)
laplacian_abs = cv2.convertScaleAbs(laplacian)  # Convert to absolute values

# Apply Sobel filters (First-order derivative for edge detection)
sobel_x = cv2.Sobel(img_gray, cv2.CV_64F, dx=1, dy=0, ksize=3)  # Vertical edges
sobel_y = cv2.Sobel(img_gray, cv2.CV_64F, dx=0, dy=1, ksize=3)  # Horizontal edges

# Compute absolute gradients
abs_sobel_x = cv2.convertScaleAbs(sobel_x)
abs_sobel_y = cv2.convertScaleAbs(sobel_y)

# Compute Gradient Magnitude (Combine Sobel X and Y)
grad_magnitude = cv2.addWeighted(abs_sobel_x, 0.5, abs_sobel_y, 0.5, 0)

# Display results side by side
plt.figure(figsize=(20, 10))

# Original Grayscale Image
plt.subplot(1, 3, 1)
plt.imshow(img_gray, cmap="gray")
plt.title("Original Grayscale Image", fontsize=18)
plt.axis("off")

# Laplacian Edge Detection
plt.subplot(1, 3, 2)
plt.imshow(laplacian_abs, cmap="gray")
plt.title("Laplacian Edge Detection", fontsize=18)
plt.axis("off")

# Sobel Gradient Magnitude
plt.subplot(1, 3, 3)
plt.imshow(grad_magnitude, cmap="gray")
plt.title("Sobel Edge Detection", fontsize=18)
plt.axis("off")

plt.tight_layout()
plt.show()
```

As we can see in the ouput we have the input image in grayscale before applying filters with the used of the Laplacian and Sobel application we did see that Laplacian detects edges in all directions at once but is more sensitive to noise. While Sobel detects edges directionally (X/Y) and gives smoother results. This make sense that laplacian application has a more detailed but noisier edges and Sobel has a more like of the directional edges (look at the plain grass picture comapre tot that of laplacian) and it is a bit more smoother.
